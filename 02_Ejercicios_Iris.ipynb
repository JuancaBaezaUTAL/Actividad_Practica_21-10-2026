## Ejercicio 1. Explorar el dataset

a) Muestra las primeras cinco filas. b) Identifica las dimensiones. c) Consulta los tipos de columna y sus valores no nulos. d) Encuentra cuáles especies aparecen en la columna `species`.

**Pista:** Recuerda `head`, `shape`, `info` y `unique`. Debes encontrar **3 especies**.


```python
import pandas as pd

# Cargar el dataset
df = pd.read_csv("iris_con_faltantes.csv")

print("a) Primeras cinco filas")
display(df.head(5))

print("b) dimensiones (filas, columnas)")
print("Dimensiones:", df.shape)

print("c) Tipos de columna y sus valores no nulos")
df.info()

print("d) Especies")
print("Especies:", df["species"].unique())
```

## Ejercicio 3. Filtrar con una o dos condiciones

a) Selecciona solamente flores de la especie `setosa`. b) Selecciona flores de `virginica` cuyo `petal_length_cm` sea **mayor que 5.5**. c) Muestra cuántas filas tiene cada resultado.

**Pista:** Utiliza una máscara de comparación y `&`; cada condición debe ir entre paréntesis.


```python
print("a) Flores unicamente de la especie 'setosa'")

mascara_setosa = df["species"] == "setosa"
setosa_df = df.loc[mascara_setosa]

print("b) Flores de 'virginica' con petal_length_cm mayor a 5.5")
mascara_virginica = (df["species"] == "virginica") & (df["petal_length_cm"] > 5.5)
virginica_grandes = df.loc[mascara_virginica]

print("c) Mostrar cuántas filas tiene cada resultado")
print("Cantidad de flores 'setosa':", len(setosa_df))
print("Cantidad de flores 'virginica' con pétalo > 5.5 cm:", len(virginica_grandes))

print("Visualizacion de muestra para comprobar")
display(setosa_df.head(3))
display(virginica_grandes.head(3))
```

## Ejercicio 4. Contar y agrupar por especie

a) Cuenta cuántas flores hay de cada especie. b) Calcula el **promedio de `petal_length_cm` para cada especie**. c) Identifica cuál tiene el promedio mayor.

**Pista:** Busca `value_counts`, `groupby` y `mean`. La columna `petal_length_cm` no contiene faltantes en esta copia.


```python
print("a)Conteo de flores por especie:")
print(df["species"].value_counts())


promedios_petalo = df.groupby("species")["petal_length_cm"].mean()
print("b)Promedio de longitud de pétalo (cm) por especie:")
print(promedios_petalo)


especie_mayor = promedios_petalo.idxmax()
valor_mayor = promedios_petalo.max()

print(f"C)La especie con el promedio mayor es '{especie_mayor}' con {valor_mayor} cm.")
```

## Ejercicio 5. Limpiar valores faltantes y quitar columnas

a) Cuenta cuántos valores faltan por columna. b) Crea **una copia** sin las filas que tengan valores vacíos y cuenta cuántas quedan. c) Crea otra copia donde rellenes únicamente los faltantes de `sepal_width_cm` con su mediana. d) Crea una tercera copia sin la columna `petal_width_cm`. **No modifiques `iris` directamente.**

**Pista:** Recuerda `isna().sum()`, `dropna()`, `fillna()`, `median()` y `drop(columns=[...])`. Esperamos **142 filas completas** antes de rellenar.


```python
print("a) Faltantes por columna:")
print(df.isna().sum())

sin_vacios = df.dropna().copy()
print("b) Filas restantes sin valores vacíos:", len(sin_vacios))

copia_rellenada = df.copy()
mediana_sepal_width = copia_rellenada["sepal_width_cm"].median()
copia_rellenada["sepal_width_cm"] = copia_rellenada["sepal_width_cm"].fillna(mediana_sepal_width)
print("c) Faltantes en 'sepal_width_cm':", copia_rellenada["sepal_width_cm"].isna().sum())

# d) Copia sin la columna 'petal_width_cm'
copia_sin_columna = df.drop(columns=["petal_width_cm"]).copy()
print("d) Columnas restantes tras drop:", list(copia_sin_columna.columns))
print("Dimensiones de la tabla sin la columna:", copia_sin_columna.shape)
```

## Ejercicio 6. Transformar columnas

a) Crea en una copia una nueva columna `petal_length_mm` convirtiendo centímetros a milímetros. b) Crea otra columna `species_codigo` asignando `S`, `V` y `G` a las tres especies. c) Comprueba cinco filas.

**Pista:** Para la conversión basta una operación sobre toda la columna; para códigos puedes usar `.map()` con un diccionario. Si quieres, prueba también `.apply(lambda ...)`.


```python
# copia independiente para trabajar las transformaciones
copia_transformada = df.copy()

# a) Nueva columna petal_length_mm convirtiendo cm a mm (* 10)
copia_transformada["petal_length_mm"] = copia_transformada["petal_length_cm"] * 10


# b) Nueva columna species_codigo asignando 'S', 'V' y 'G'
mapeo_especies = {
    "setosa": "S",
    "versicolor": "V",
    "virginica": "G"
}
copia_transformada["species_codigo"] = copia_transformada["species"].map(mapeo_especies)

# c) Comprobar cinco filas con las columnas clave
columnas_revision = ["species", "species_codigo", "petal_length_cm", "petal_length_mm"]
display(copia_transformada[columnas_revision].head(5))
```

## Ejercicio 9. Relacionar tablas con `merge`
Crea una tabla pequeña con las columnas `species` y `codigo`: setosa→S, versicolor→V y virginica→G. Une esta tabla con `iris` mediante la especie. Verifica que aparecen 150 filas y una columna adicional. Los códigos son solamente etiquetas didácticas.

**Pista:** Puedes construir la tabla auxiliar con `pd.DataFrame({...})` y hacer un `merge` por `species`.


```python
# 1 Crear la tabla auxiliar con códigos
codigos_iris = pd.DataFrame({
    "species": ["setosa", "versicolor", "virginica"],
    "codigo": ["S", "V", "G"]
})

print("Tabla auxiliar:")
print(codigos_iris)

# 2 Relacionar la tabla auxiliar con iris mediante la columna clave común
iris_combinado = df.merge(codigos_iris, on="species", how="left")

# 3 Comprobar dimensiones y columnas resultantes
print("Dimensiones originales de df:", df.shape)
print("Dimensiones tras merge:", iris_combinado.shape)

# 4. Visualizar las primeras filas para confirmar la nueva columna
display(iris_combinado[["species", "codigo", "petal_length_cm", "petal_width_cm"]].head(5))
```
